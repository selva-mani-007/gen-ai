```python
# Import required libraries
import gradio as gr
from transformers import pipeline

# Load a pre-trained sentiment analysis model from Hugging Face
sentiment_model = pipeline("sentiment-analysis")

# Define the function to analyze sentiment
def analyze_sentiment(text):
    result = sentiment_model(text)[0]
    label = result['label']
    score = result['score']
    return f"Sentiment: {label}, Confidence: {score:.2f}"

# Create the Gradio interface
description = "This is an AI-powered sentiment analysis tool that analyzes customer feedback and determines the sentiment (positive, neutral, or negative) with confidence scores."
app = gr.Interface(
    fn=analyze_sentiment,
    inputs=gr.Textbox(lines=2, placeholder="Enter customer feedback here..."),
    outputs="text",
    title="Customer Feedback Sentiment Analyzer",
    description=description,
)

# Launch the app
app.launch()
```

```python


# Import required libraries
import gradio as gr
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import make_pipeline

# Load the SMS Spam Collection dataset
url = "https://raw.githubusercontent.com/mohitgupta-omg/Kaggle-SMS-Spam-Collection-Dataset-/master/spam.csv"
data = pd.read_csv(url, encoding='latin-1')

# Preprocess the dataset
data = data[['v1', 'v2']]
data.columns = ['label', 'text']
data['label'] = data['label'].map({'ham': 0, 'spam': 1})

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(data['text'], data['label'], test_size=0.2, random_state=42)

# Create a text classification pipeline using Naive Bayes
model = make_pipeline(CountVectorizer(), MultinomialNB())
model.fit(X_train, y_train)

# Define the prediction function
def predict_spam(text):
    prediction = model.predict([text])[0]
    return "Spam" if prediction == 1 else "Not Spam"

# Create a Gradio interface
app = gr.Interface(
    fn=predict_spam,
    inputs=gr.Textbox(lines=2, placeholder="Enter your message here..."),
    outputs="text",
    title="Spam Detection App",
    description="Classify SMS messages as Spam or Not Spam using machine learning."
)

# Launch the app
app.launch()
```

```python


import gradio as gr
from transformers import pipeline

# Load the summarization pipeline
summarizer = pipeline("summarization")

def summarize_text(text):
    summary = summarizer(text, max_length=130, min_length=30, do_sample=False)
    return summary[0]['summary_text']

# Create Gradio interface
interface = gr.Interface(
    fn=summarize_text,
    inputs=gr.Textbox(label="Input Text", placeholder="Paste your text here..."),
    outputs=gr.Textbox(label="Summary"),
    examples=[
        ["The Eiffel Tower is one of the most recognizable structures in the world. Located in Paris, France, it was completed in 1889 and has become a global cultural icon of France."],
        ["Artificial Intelligence (AI) refers to the simulation of human intelligence in machines that are programmed to think like humans and mimic their actions."]
    ],
    title="Extractive Text Summarization",
    description="This tool summarizes your text using an extractive summarization model."
)

interface.launch(share=True)
```

```python


import gradio as gr
from transformers import pipeline

# Load a pre-trained text generation model (e.g., GPT-2)
generator = pipeline("text-generation", model="gpt2")

def generate_paragraph(prompt, max_length=150):
    """
    Generate a paragraph based on the given prompt.
    """
    result = generator(prompt, max_length=max_length, num_return_sequences=1, do_sample=True)
    return result[0]["generated_text"]

# Create Gradio interface
interface = gr.Interface(
    fn=generate_paragraph,
    inputs=[
        gr.Textbox(label="Prompt", placeholder="Enter a topic or starting sentence..."),
        gr.Slider(label="Max Length", minimum=50, maximum=300, step=10, value=150)
    ],
    outputs=gr.Textbox(label="Generated Paragraph"),
    title="Text Paragraph Generator",
    description="Generate well-structured paragraphs based on your input prompt using GPT-2.",
    examples=[
        ["The importance of artificial intelligence in modern society"],
        ["Climate change and its impact on global ecosystems"],
        ["The future of space exploration and colonization"]
    ]
)

interface.launch(share=True)
```

```python


import gradio as gr
from sentence_transformers import SentenceTransformer, util
import torch

# Load a pre-trained model for embeddings (e.g., all-MiniLM-L6-v2)
model = SentenceTransformer('all-MiniLM-L6-v2')

# Example document collection
documents = [
    "The Eiffel Tower is located in Paris, France. It was constructed in 1889.",
    "Python is an interpreted, high-level programming language.",
    "Artificial Intelligence (AI) refers to the simulation of human intelligence in machines.",
    "Climate change is impacting global ecosystems and causing rising sea levels.",
    "The future of space exploration involves colonization of Mars and beyond."
]

# Precompute document embeddings
document_embeddings = model.encode(documents, convert_to_tensor=True)

def retrieve_information(query, top_k=3):
    """
    Retrieve the most relevant documents based on the user's query.
    """
    # Encode the query into an embedding
    query_embedding = model.encode(query, convert_to_tensor=True)

    # Compute cosine similarity between query and document embeddings
    similarities = util.pytorch_cos_sim(query_embedding, document_embeddings)

    # Get the top-k most relevant documents
    top_results = torch.topk(similarities, k=top_k)

    # Format the results for display
    results = []
    for score, idx in zip(top_results[0][0], top_results[1][0]):
        results.append(f"Document: {documents[idx]}\nScore: {score.item():.4f}")

    return "\n\n".join(results)

# Create Gradio interface
interface = gr.Interface(
    fn=retrieve_information,
    inputs=[
        gr.Textbox(label="Query", placeholder="Enter your question or search query..."),
        gr.Slider(label="Number of Results", minimum=1, maximum=5, step=1, value=3)
    ],
    outputs=gr.Textbox(label="Relevant Documents"),
    title="Information Retrieval System",
    description="Retrieve relevant documents based on your query using semantic search.",
    examples=[
        ["Where is the Eiffel Tower located?", 3],
        ["What is artificial intelligence?", 2],
        ["What are the effects of climate change?", 3]
    ]
)

interface.launch(share=True)
```

```python


import gradio as gr
from google.cloud import texttospeech

# Set up Google Cloud credentials (replace with your JSON key file path)
import os
os.environ["GOOGLE_APPLICATION_CREDENTIALS"] = "path/to/your-service-account-key.json"

# Initialize the Text-to-Speech client
client = texttospeech.TextToSpeechClient()

def text_to_speech(text, language_code="en-US", voice_name="en-US-Wavenet-D", speaking_rate=1.0):
    """
    Convert input text to speech using Google Cloud Text-to-Speech API.
    """
    # Set the text input
    synthesis_input = texttospeech.SynthesisInput(text=text)

    # Configure the voice settings
    voice = texttospeech.VoiceSelectionParams(
        language_code=language_code,
        name=voice_name,
        ssml_gender=texttospeech.SsmlVoiceGender.NEUTRAL
    )

    # Configure the audio settings
    audio_config = texttospeech.AudioConfig(
        audio_encoding=texttospeech.AudioEncoding.MP3,
        speaking_rate=speaking_rate
    )

    # Perform the text-to-speech request
    response = client.synthesize_speech(
        input=synthesis_input,
        voice=voice,
        audio_config=audio_config
    )

    # Save the output audio file
    output_path = "output.mp3"
    with open(output_path, "wb") as out:
        out.write(response.audio_content)

    return output_path

# Create Gradio interface
interface = gr.Interface(
    fn=text_to_speech,
    inputs=[
        gr.Textbox(label="Enter Text", placeholder="Type something to convert to speech..."),
        gr.Dropdown(choices=["en-US", "en-GB", "es-ES"], label="Language Code", value="en-US"),
        gr.Dropdown(choices=["en-US-Wavenet-D", "en-US-Wavenet-A"], label="Voice Name", value="en-US-Wavenet-D"),
        gr.Slider(label="Speaking Rate", minimum=0.5, maximum=2.0, step=0.1, value=1.0)
    ],
    outputs=gr.Audio(label="Generated Speech"),
    title="Google Cloud Text-to-Speech",
    description="Convert any text into speech using Google Cloud TTS API.",
    examples=[
        ["Hello, how are you?", "en-US", "en-US-Wavenet-D", 1.0],
        ["This is an example of text-to-speech conversion.", "en-GB", "en-US-Wavenet-A", 1.2],
        ["Gradio makes it easy to create interactive applications.", "es-ES", "en-US-Wavenet-D", 0.8]
    ]
)

# Launch the Gradio app
interface.launch(share=True)
```

```python
import gradio as gr
import random

# Function to handle chatbot responses
def respond(message, chat_history):
    bot_message = random.choice(["Hello!", "How can I assist you?", "Have a great day!"])
    chat_history.append({"role": "user", "content": message})
    chat_history.append({"role": "assistant", "content": bot_message})
    return "", chat_history

# Create the Gradio interface
with gr.Blocks() as demo:
    chatbot = gr.Chatbot(type="messages")
    msg = gr.Textbox(label="Your Message")
    clear = gr.ClearButton([msg, chatbot])

    # Link the response function to the input box
    msg.submit(respond, [msg, chatbot], [msg, chatbot])

# Launch the app
demo.launch(share=True)
```

```python
from transformers import AutoModelForCausalLM, AutoTokenizer

# Load GPT-2 model and tokenizer
model_name = "gpt2"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

def gpt2_chatbot(input_text, chat_history=[]):
    # Append user input to chat history
    chat_history.append(f"User: {input_text}")

    # Prepare input for GPT-2
    input_ids = tokenizer.encode(" ".join(chat_history), return_tensors="pt")

    # Generate response
    output = model.generate(input_ids, max_length=150, num_return_sequences=1, pad_token_id=tokenizer.eos_token_id)
    response = tokenizer.decode(output[:, input_ids.shape[-1]:][0], skip_special_tokens=True)

    # Append bot response to chat history
    chat_history.append(f"Bot: {response}")

    # Return the updated chat history
    return response, chat_history


import gradio as gr

# Define the Gradio interface
with gr.Blocks() as demo:
    chatbot = gr.Chatbot(label="GPT-2 Chatbot")
    user_input = gr.Textbox(label="Your Message")
    clear_button = gr.ClearButton([user_input, chatbot])

    def respond(input_text, chat_history=[]):
        response, updated_history = gpt2_chatbot(input_text, chat_history)
        return "", updated_history

    user_input.submit(respond, [user_input, chatbot], [user_input, chatbot])

# Launch the Gradio app
demo.launch(share=True)
```
